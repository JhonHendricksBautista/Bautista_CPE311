Remove characters from the lyrics

```python
import re

newline = lambda x: re.sub('\n', ' ', x) # remove \n
alphanumeric = lambda x: re.sub('\w*\d\w*', ' ', x) # remove alphanumeric words

corpus = beatles.text.map(newline).map(alphanumeric)
corpus.head()
```

Create a document term matrix using Count Vectorizer with the stop words turned on to English

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

cv = CountVectorizer(stop_words="english")
X = cv.fit_transform(corpus).toarray()

dt = pd.DataFrame(X, columns=cv.get_feature_names_out())
dt = dt.set_index(beatles.song)

dt.head()
```

Define the cosine similarity calculation

```python
from numpy import dot
from numpy.linalg import norm

cosine = lambda v1, v2: dot(v1, v2) / (norm(v1) * norm(v2))
```

Calculate all similarities and sort by the most similar

```python
similarity = [cosine(imagine, song) for song in X]
sorted(list(zip(similarity, beatles.song)), reverse=True)[1:6]
```

## Use a dataset that consists of text document

```python
#type your code here
import pandas as pd
import re
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from tqdm import tqdm

df = pd.read_csv('/content/drive/MyDrive/datasci3/text_class.csv')
df.head(20)
```

## PREPROCESSING

Tokenize (words)


```python
# Tokenize words (just split by whitespace)
df['tokens_basic'] = df['Text'].apply(lambda x: x.split())
df['tokens_basic'].head()
```

Tokenize (words with remove punctuation) / Tokenize (words with remove special characters and numbers)


```python
# Only letters, remove numbers and special characters
df['tokens_clean'] = df['Text'].apply(lambda x: re.findall(r'\b[a-zA-Z]+\b', x))
df['tokens_clean'].head()
```

Tokenize (words with remove punctuation and all lowercase)


```python
# Only letters, remove numbers and special characters
df['tokens_clean'] = df['Text'].apply(lambda x: re.findall(r'\b[a-zA-Z]+\b', x))
df['tokens_clean'].head()
```

Tokenize (words with remove punctuation, all lowercase and remove stop words)


```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Remove stop words
df['tokens_no_stop'] = df['Text'].apply(
    lambda x: [w for w in re.findall(r'\b[a-zA-Z]+\b', x.lower()) if w not in ENGLISH_STOP_WORDS]
)
df['tokens_no_stop'].head()
```

Tokenize (words with remove punctuation, all lowercase and  remove stop words, with speech_tagging)

```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets')

from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
import re

def tokenize_pos(text):
    words = [w for w in re.findall(r'\b[a-zA-Z]+\b', text.lower()) if w not in ENGLISH_STOP_WORDS]
    return nltk.pos_tag(words)

df['tokens_pos'] = df['Text'].apply(tokenize_pos)
df['tokens_pos'].head()
```

***In the different tokenization process it transforms the text into different bits and keeps only the important or relevent text data. This lessens the noise and even computational strain when performing analysis or training on the text data. With tokenizaion we are able to extract important text data only.***
